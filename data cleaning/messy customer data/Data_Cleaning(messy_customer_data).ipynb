```python
#Installing the libraries for data cleaning
import pandas as pd
import numpy as np
import regex as re
```

```python
#Loading the data
data = pd.read_csv('messy_customer_orders.csv')
data
```

```python
#Making an copy of the data for disaster management
data_copy = data
```

```python
#Tracking the number of null values are there in each columns
data_copy.isnull().sum()
```

```python
#Tracking for the duplicate values
duplicate_values = data_copy.duplicated().sum()
print("Count of Duplicated Values are : ",duplicate_values)
data_copy[data_copy.duplicated(keep=False)].sort_values('order_id').head(10)
```

```python
#Droping the Duplicated values
data_copy = data_copy.drop_duplicates()
```

```python
#Now checking duplicates values and the shape of the dataset after removing the duplicate values
data_shape = data_copy.shape
print("The Shape of the data after removing duplicates : ",data_shape)
data_copy[data_copy.duplicated(keep=False)].sort_values('order_id').head(10)
```

```python
data_copy
```

```python
#Now replacing the null values of total amount using median method
data_copy['total_amount'] = data_copy['total_amount'].fillna(data_copy['total_amount'].median())
data_copy
```

```python
for i in range(0,15):
  print(data_copy['total_amount'][i])
```

```python
#let go for replacing the unit price null values

#firstly convert the object data type to string
data_copy['unit_price'] = data_copy['unit_price'].astype(str)

#secondly remove the special characters from the string
data_copy['unit_price'] = data_copy['unit_price'].str.replace(r'[^0-9.]','',regex=True)

#now convert the string column to float64
data_copy['unit_price'] = pd.to_numeric(data_copy['unit_price'],errors='coerce').astype('float64')

#finally replace the null values
data_copy['unit_price'] = data_copy['unit_price'].fillna(data_copy['unit_price'].median())
data_copy
```

```python
data_copy.isnull().sum()
```

```python
#Now filling the remaining categorical columns
cate_col = ['customer_id','country','category']
data_copy[cate_col]  = data_copy[cate_col].fillna('unknown')
data_copy
```
