```python
#making an copy of the dataset
df = data.copy()
df.shape
```

```python
#converting blank spaces null by making it float datatype
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].str.strip(), errors='coerce')
```

```python
#Assigning 0 to totalcharges of the new customers whose tenure is zero
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
#converting churn column data into numerical data (1 or 0)
df['Churn_Numeric'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)
```

```python
#grouping different tenure group
def tenure_cohort(tenure):
    if tenure <= 12:
        return "0-1 Year (High Risk)"
    elif tenure <= 24:
        return "1-2 Years"
    elif tenure <= 48:
        return "2-4 Years"
    else:
        return "4+ Years (Loyal)"

df['Tenure_Cohort'] = df['tenure'].apply(tenure_cohort)
```

```python
#tech support availability group
df['Has_TechSupport'] = df['TechSupport'].apply(lambda x: 'Yes' if x == 'Yes' else 'No / No Internet')
```

```python
df.to_csv("Update_customer_churn.csv",index=False)
```
