```python
#making an copy of the dataset
df = data.copy()
df.shape
```

```python
#removing the duplicates from the dataset
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
# There are two approachs
# first is to delete the rows which has null values
# second is to fill those using fillna() method
# But the rows which having the null values are not that important in dashboard making so i go for first approach

#Deleting all the null values rows
df = df.dropna(subset=['Description','CustomerID'])
```

```python
df.shape
```

```python
#deleting rows which has negative/0 quautity and unit prices
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]
```

```python
#converting UK pounds to indian rupees
exchange_rate = 127.60
df['UnitPrice_INR'] = df['UnitPrice'] * exchange_rate
```

```python
df.isnull().sum()
```

```python
#extracting dates from the InvoiceDate column
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')

df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['Date'] = df['InvoiceDate'].dt.date
df['Time'] = df['InvoiceDate'].dt.time
```

```python
df.to_csv("Update_online_retail.csv",index=False)
```
